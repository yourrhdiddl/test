```python
# 1. 각 상품의 정가 입력 받아서 price에 저장
onion_price = int(input("양파 정가 입력(원): "))
tomato_price = int(input("토마토 정가 입력(원): "))
potato_price = int(input("감자 정가 입력(원): "))

# 2. 부가세 10%를 더한 금액 계산
onion_total = onion_price * 1.1
tomato_total = tomato_price * 1.1
potato_total = potato_price * 1.1

# 3. 부가세 포함 메뉴판 출력
print("\n" + "="*30)
print("            Menu")
print("     (부가세 포함 가격)")
print("="*30)
print("1. 양파  :", int(onion_total), "원")
print("2. 토마토  :", int(tomato_total), "원")
print("3. 감자  :", int(potato_total), "원")
print("="*30 + "\n")

onion_pcs = int(input("양파 개수: "))
tomato_pcs = int(input("토마토 개수: "))
potato_pcs = int(input("감자 개수: "))

total_price = onion_pcs*onion_total + tomato_pcs*tomato_total + potato_pcs*potato_total

print("="*30)
print("총 가격  :", int(total_price), "원")
print("="*30)
```
